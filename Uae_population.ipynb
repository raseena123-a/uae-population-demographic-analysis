```python
import pandas as pd
df = pd.read_csv("API_ARE_DS2_en_csv_v2_11310.csv", skiprows=4)
df.head()
```

```python
df.head()
```

```python
df.info()
```

```python
df_uae = df[df['Country Name'] == 'United Arab Emirates']
```

```python
df_uae

```

```python
df_uae = df_uae.drop(columns=[
    'Country Code',
    'Indicator Name',
    'Indicator Code'
])
```

```python
df_long = df_uae.melt(
    id_vars=['Country Name'],
    var_name='Year',
    value_name='Population'
)
```

```python
df_long['Year'] = pd.to_numeric(df_long['Year'], errors='coerce')
df_long['Population'] = pd.to_numeric(df_long['Population'], errors='coerce')
```

```python
df_clean = df_long.dropna()
```

```python
df_clean = df_clean.sort_values(by='Year')
```

```python
df_clean.head()
df_clean.tail()
df_clean.info()
```

```python
df_clean.to_csv("uae_population_clean.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("uae_population_clean.csv")
df.head()
```

```python
#Latest population
latest_year = df['Year'].max()
latest_population = df[df['Year'] == latest_year]['Population'].values[0]

latest_year, latest_population
```

```python
#Population growth trend
df['Growth'] = df['Population'].diff()
df.head()
```

```python
#Growth rate (%)
df['Growth_Rate_%'] = df['Population'].pct_change() * 100
df.head()
```

```python
#Average growth rate
df['Growth_Rate_%'].mean()
```

```python
#Create first visualization
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df['Year'], df['Population'])
plt.xlabel("Year")
plt.ylabel("Population")
plt.title("UAE Population Growth Trend")
plt.show()
```

```python
df.to_csv("uae_population_analysis_ready.csv", index=False)
```
